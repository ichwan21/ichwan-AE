```python
import os

print(os.getcwd())
print(os.listdir('/content'))

import pandas as pd


filename = 'Estate-Agents-Purchased.csv'
data_Estate = pd.read_csv(filename, encoding='cp1252')
data_Estate
data_Estate.columns

filename_2 = 'UK-Agents-Validated.csv'
data_UK_Agent = pd.read_csv(filename_2, encoding='cp1252')
data_UK_Agent

data_Estate.columns
data_UK_Agent.columns


#Merge Tables
merged_df = pd.merge(
    data_Estate[['POSTCODE','FIRST NAME', 'LAST NAME', 'WEBSITE ADDRESS', 'EMAIL']],
    data_UK_Agent[['POSTCODE', 'PHONE']],
    on='POSTCODE',
    how='left'
)
merged_df

#We need a consolidated list of emails address, agentt names and addresses, with telephone numbers if possible.

#Merge Tables
#merged_df = pd.merge(
#    df1[["column_join", "column_name"]],
#    df2[["column_join", column_name]],
#    on='column_join',
#    how='{joinType (inner/outer/left/right)}'
#)

```
