```python
#what does this endpoint provide?
#what is the structure of the returned JSON? Line_status is a list
#How can the response be turned into information that answers a useful question?


#Importing Libraries
import pandas as pd
import requests       #used to send HTTP request to website (communicate with the website through APIs)

#Display all columns when wiewing a pandas DataFrame
pd.set_option('Display.max_columns', None)

BASE_URL = 'https://api.tfl.gov.uk'

#Check that the API is available
response = requests.get(
    f"{BASE_URL}/Line/Mode/tube/Status",    #f string to combine variables BASE_URL with text
    timeout=30
)

print("Status code:", response.status_code)
response.raise_for_status()

line_status = response.json()
print("Response type:", type(line_status).__name__)  #Its a List
print("Lines returned:", len(line_status))


#2. Inspect and parse the JSON response
#2.1 Inspect the overall structure

print("Overall type:", type(line_status).__name__)
print("Number of line records:", len(line_status))

first_line = line_status[0]
print("First item type:", type(first_line).__name__)
print("Top-level keys:")
print(first_line.keys())

######################
df = pd.DataFrame(line_status)
print(df)
######################



```
